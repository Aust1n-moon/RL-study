# What is RL?
- Trying to understand a way of decision making. In engineering, mostly used in optimal control

# How is RL different from other ML?
- no supervisor, only a reward signal
- Feedback is delayd, not instantaneous - can only see if it was a good/bad decision after many steps
- Time really matters (sequential, non i.i.d data - independent and identically distributed)
- Agent's actions affect the subsequent data it receives

    **examples**
    - fly stunt manoeuvres in a helicopter - you want a helicopter to make a manoeuvre without someone telling it to do it and you can only tell good or bad afterwards
        - Reward: + for desired trajectory, - for crashing
    - managing an investment portfolio - decision made in real time and you can only reward it after. 
        - positive reward for each $ in bank account

# Rewards
- a reward Rt is a scalar feedback signal
- tells how wwell an agent is doing at step t and the agent's job is to maximize the culmulative reward
- RL is based on reward hypothesis: all goals can be described by the maximization of expected culmulative reward

Goal is to select actionis to maximize total future reward
- actions may cause long term consequences
- may sacrifice immediate reward to gain more long-term reward
    - ex. refuleling a helicopter to prevent crash in a several hour

# Agent and Environment
**agent**
- Observation $ O_t $, Reward $ R_t $, Action $ A_t $
- Gets an observation coming in, then from the reward that comes in from that observation, must make a decision as an action
- A this is a loop
**environment**
- vice versa

# History
What the agent has seen so far: $ H_t = A_0, O_0, R_0, ..., A_t, O_t, R_t $ (all observable variables up to time t)
- what happens next depends on this history:
    - agent selects actions
    - the environment selects observations/rewards

# State #
- information used to determine what happens next $ S_t = f(H_t) $

**environment state**
- environment state $ S_t^e $ is the environment's private representation (i.e. whatever data the env. uses to pick the next O and R)
- this is not usually visible to the agent

**agent state**
- agent state $ S_t^a $ is the agent's internal representation (i.e. whatever information the agent sues to pick the next action, info used by RL algorithms)
- can be any function of history: $ S_t^a = f(H_t) $

# Information (Markov State)
Contains all useful information from history - summarizing everything about history that matters to predict the future
- a state $ S_t $ is Markov if and only if $ P[S_{t+1} | S_t] = P[S_{t+1} | S_1, ..., S_t] $
- This means that the future is independent of the past given the present: $ H_1:t -> S_t -> H_{t+1: \infty} $
- Only need to store S and if its Markov, you can get rid of all the history as it wouldn't have any effect on the future
- The environment state and history is Markov

**fully observable environments**
Agent directly observes environment state
- $ O_t = S_t^a = S_t^e $
- formally, this is a Markov decision process (MDP)

**partially observable environments**
Agent indirectly observes environment
    - a robot camera vision isn't told its absolute location
    - trading agent only observes current price
- now agent state is not environment state
- known as partially observable Markov decision process(POMDP)
- agent must construct its own state represetation $ S_t^a $
    - ex. complete history, believs of environmenet state, recurrent neural network (RNN)


# Inside an RL agent
RL agent may include one or more of these componenets:
- policy: agent's behaviour function
    - agent's behaviour - what it will do at each state
    - map from state to action 
        - ex. Deterministic policy
        - ex. Stochastic policy
- value function: how good each state and/ or action - ex. how much steps are left from the goal in a maze
    - prediction of future reward - used to evaluate goodness/badness of states
    - therhefore to select between actions
        - $ V_\pi(s) = E_\pi[R_t + \gamma*R_t + \gamma^2*R_t + ... | S_t = s] $
- model: agent's representation of the environment - ex. building its own map in an environment, like labelling all paths it went through as -1
    - predict waht the environment will do next
    - transitions: P predicts the next state
    - rewards: R predicts the next reward (immediate reward)

**categorizing RL agents**
- value based
    - no policy (implicit) - just look at value functions and pick out the best depending on value function
    - value function
- policy based
    - policy - like the arrows in each state of a maze, telling it where to go at each state rather than having an reward function and having to choose 
    - no value vunction
- active critic
    - policy
    - value function
- model free
    - policy and/or value function
    - no model
- model based
    - policy and/or value function
    - model
